# Saving Calculations

This is a full working example of how to save single Sample or BatchFile calculations to a CSV or Excel file.

```python
import VESIcal as v
```

## Save utility

A commonly requested feature is the ability to save out computed values that are not in the form of pandas dataframes. For example, a saturation pressure calculation might return a scalar, but you want to easily export that result to your machine for other analysis. VESIcal now supports this, allowing the user to specify one or more python objects to save to a csv or xlsx file and optionally include descriptors of those values.

The `save_results()` VESIcal method takes in standard python objects, so you can pass any supported object type, like this:



```python
# Save a dict, using keys as the descriptions
single_dict = {"H2O": 1, "CO2": 2}
v.save_results("single_dict.csv", single_dict)

# save a list of numbers with no description
v.save_results("list_of_numbers.csv", [1, 2, 3])

# save a list of numbers with a description of each number
v.save_results("list_of_numbers_desc.csv", [1, 2, 3], descriptions=["One", "Two", "Three"])

# save a single value
v.save_results("float.csv", 5)
```

## Single Sample
This example creates a Sample, calculates the saturation pressure, and then exports the result to a CSV file and to an Excel file.

```python
mysample = v.Sample({
    'SiO2': 77.5,
    'TiO2': 0.08,
    'Al2O3': 12.5,
    'Fe2O3': 0.207,
    'Cr2O3': 0.0,
    'FeO': 0.473,
    'MnO': 0.0,
    'MgO': 0.03,
    'NiO': 0.0,
    'CoO': 0.0,
    'CaO': 0.43,
    'Na2O': 3.98,
    'K2O': 4.88,
    'P2O5': 0.0,
    'H2O': 5.5,
    'CO2': 0.05
})

satP = v.calculate_saturation_pressure(sample=mysample, temperature=1200)
diss = v.calculate_dissolved_volatiles(sample=mysample, pressure=1000, temperature=1200)
degas = v.calculate_degassing_path(sample=mysample, temperature=1200)
```

```python
# save saturation pressure calculation (single number) with a description, which we set as our sample name
v.save_results("satP.csv", satP, descriptions=["my sample name"])

# save the dissolved volatiles (H2O and CO2 values) with a description
v.save_results("diss.csv", diss, descriptions=["my sample name"])

# save the degassing results without a description
v.save_results("degas.csv", degas)
```

## BatchFile
This example creates a BatchFile by importing an Excel file, calculates the saturation pressure of all samples, and then exports the result to a CSV file and to an Excel file.

```python
myfile = v.BatchFile('../example_files/example_data.xlsx')

# calculate sat P using temperatures from the imported excel file
satPs_batch = myfile.calculate_saturation_pressure(temperature="Temp")

myfile.save_excel(filename="satPs_batch.xlsx", calculations=[satPs_batch])
myfile.save_csv(filename="satPs_batch.csv", calculations=[satPs_batch])
```
